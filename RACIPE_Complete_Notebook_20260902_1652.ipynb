## 1. 📦 Imports and Setup

```python
import numpy as np\nimport pandas as pd\nimport matplotlib.pyplot as plt\nimport seaborn as sns\nfrom scipy import stats\nfrom sklearn.decomposition import PCA\nimport warnings\nwarnings.filterwarnings('ignore')\n\nprint('✅ All libraries imported successfully!')
```

## 2. 📊 Load Data

```python
# RACIPE results\nracipe_medians = pd.read_csv('racipe_10000_medians.csv')\nracipe_scores = pd.read_csv('racipe_10000_emt_scores.csv')['emt_score']\n\n# GSE69667 real data\nreal_data = pd.read_csv('GSE69667_EMT_scores_real.csv')\n\nprint(f'✅ RACIPE: {len(racipe_scores)} models')\nprint(f'✅ GSE69667: {len(real_data)} time points')\nprint(f'   Time points: {real_data["Time"].values}')
```

## 3. 📊 Statistical Summary

```python
print('='*70)\nprint(' RACIPE STATISTICS')\nprint('='*70)\nprint(f'Models: {len(racipe_scores)}')\nprint(f'EMT Score - Mean: {racipe_scores.mean():.3f} ± {racipe_scores.std():.3f}')\nprint(f'EMT Score - Range: {racipe_scores.min():.3f} - {racipe_scores.max():.3f}')\n\n# State distribution\ne = sum(racipe_scores < 0.3)\nh = sum((racipe_scores >= 0.3) & (racipe_scores <= 0.7))\nm = sum(racipe_scores > 0.7)\nprint(f'\nState distribution:')\nprint(f'  Epithelial:  {e/len(racipe_scores)*100:.1f}%')\nprint(f'  Hybrid:      {h/len(racipe_scores)*100:.1f}%')\nprint(f'  Mesenchymal: {m/len(racipe_scores)*100:.1f}%')\n\nprint('\n' + '='*70)\nprint(' GSE69667 STATISTICS')\nprint('='*70)\nprint(f'Time points: {len(real_data)}')\nprint(f'EMT Score - Mean: {real_data["EMT_Score"].mean():.3f} ± {real_data["EMT_Score"].std():.3f}')\nprint(f'EMT Score - Range: {real_data["EMT_Score"].min():.3f} - {real_data["EMT_Score"].max():.3f}')
```

## 4. 📈 Key Figures

```python
fig, axes = plt.subplots(1, 2, figsize=(14, 5))\n\n# EMT Score distribution\nax = axes[0]\nsns.kdeplot(racipe_scores, label='RACIPE (10k)', ax=ax, color='blue')\nsns.kdeplot(real_data['EMT_Score'], label='GSE69667', ax=ax, color='red')\nax.axvline(0.3, color='green', linestyle='--', alpha=0.7, label='E threshold')\nax.axvline(0.7, color='red', linestyle='--', alpha=0.7, label='M threshold')\nax.set_xlabel('EMT Score')\nax.set_ylabel('Density')\nax.set_title('RACIPE vs GSE69667')\nax.legend()\nax.grid(alpha=0.3)\n\n# Real trajectory\nax = axes[1]\nax.plot(real_data['Time'], real_data['EMT_Score'], 'o-', color='red', linewidth=2, markersize=8)\nax.axhline(0.3, color='green', linestyle='--', alpha=0.7, label='E threshold')\nax.axhline(0.7, color='red', linestyle='--', alpha=0.7, label='M threshold')\nax.set_xlabel('Time (hours)')\nax.set_ylabel('EMT Score')\nax.set_title('GSE69667 Trajectory (A549 + TGF-β)')\nax.legend()\nax.grid(alpha=0.3)\n\nplt.tight_layout()\nplt.savefig('figure_new_notebook.png', dpi=300)\nplt.show()\nprint('✅ Figures created!')
```
